# BTVN 01

Cho y = 5x^5 + 6x^3 - 3x + 1

Cho biết độ dốc của đa thức trên ở điểm nào

```python
import torch

# Tạo tensor x với requires_grad=True
# Thử với nhiều điểm khác nhau
diem_can_kiem_tra = [0.0, 1.0, 2.0, -1.0, 0.5]

print("Hàm số: y = 5x^5 + 6x^3 - 3x + 1")
print("Đạo hàm: dy/dx = 25x^4 + 18x^2 - 3")
print("\nĐộ dốc của đa thức tại các điểm:\n")

for diem in diem_can_kiem_tra:
    x = torch.tensor(diem, requires_grad=True)
    
    # Định nghĩa hàm số
    y = 5*x**5 + 6*x**3 - 3*x + 1
    
    # Tính gradient
    y.backward()
    
    print(f"Tại x = {x.item():6.1f}: y = {y.item():10.4f}, độ dốc dy/dx = {x.grad.item():10.4f}")

print("\n=> Độ dốc thay đổi tùy theo vị trí x")
print("=> Độ dốc = 0 khi x ≈ -0.61 hoặc x = 0 (điểm cực trị)")
```

# BTVN 02

Tạo 1 tensor ban đầu có giá trị là 2

Định nghĩa hàm số  và tính gradient

y = x^3 + 2x^2 + 5x + 1

Hãy tính đạo hàm dy/dx tại giá trị của x

Dùng phương pháp Gradient Descent với Learning Rate alpha = 0.1 để cập nhật giá trị x trong 10 vòng lặp

```python
import torch

# Tạo tensor ban đầu với giá trị 2
x = torch.tensor(2.0, requires_grad=True)

# Learning rate
alpha = 0.1

# Số vòng lặp
epochs = 10

print("Hàm số: y = x^3 + 2x^2 + 5x + 1")
print(f"Learning Rate: {alpha}")
print(f"Giá trị x ban đầu: {x.item()}\n")

for i in range(epochs):
    # Định nghĩa hàm số
    y = x**3 + 2*x**2 + 5*x + 1
    
    # Tính gradient
    y.backward()
   
    print(f"Vòng {i+1}: x = {x.item():.4f}, y = {y.item():.4f}, dy/dx = {x.grad.item():.4f}")
    
    # Cập nhật x bằng Gradient Descent: x = x - alpha * gradient
    with torch.no_grad():
        x -= alpha * x.grad
    
    # Reset gradient về 0 cho vòng lặp tiếp theo
    x.grad.zero_()

print(f"\nGiá trị x cuối cùng: {x.item():.4f}")
```

# BTVN 03

Tạo một tập dữ liệu giả lập với x là số giờ học (ngẫu nhiên từ 1-10) và y là số điểm được tính theo công thức y = 3x + 5 + noise 

Với noise là một giá trị ngẫu nhiên nhỏ

1. Khởi tạo tham số w và b ngẫu nhiên với requires_grad = True
2. Tính MSE
3. Tính Gradient
4. Cập nhật tham số w và b bằng Gradient Descent với Learning Rate alpha = 0.01
5. Lặp lại quá trình trên trong 100 vòng lặp để xem sự hội tụ của mô hình.

```python
import torch
import numpy as np

# Tạo tập dữ liệu giả lập
np.random.seed(42)
n_samples = 100

# x: số giờ học (ngẫu nhiên từ 1-10)
x_data = np.random.uniform(1, 10, n_samples)

# y = 3x + 5 + noise
noise = np.random.normal(0, 0.5, n_samples)
y_data = 3 * x_data + 5 + noise

# Chuyển sang tensor
x = torch.FloatTensor(x_data).reshape(-1, 1)
y = torch.FloatTensor(y_data).reshape(-1, 1)

# 1. Khởi tạo tham số w và b ngẫu nhiên với requires_grad=True
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

print(f"Công thức thực: y = 3x + 5 + noise")
print(f"Tham số khởi tạo: w = {w.item():.4f}, b = {b.item():.4f}\n")

# Learning rate và số epochs
alpha = 0.01
epochs = 100

for epoch in range(epochs):
    # Dự đoán: y_pred = w*x + b
    y_pred = w * x + b
    
    # 2. Tính MSE (Mean Squared Error)
    loss = ((y_pred - y) ** 2).mean()
    
    # 3. Tính Gradient
    loss.backward()
    
    # 4. Cập nhật tham số w và b bằng Gradient Descent
    with torch.no_grad():
        w -= alpha * w.grad
        b -= alpha * b.grad
        
        # Reset gradient
        w.grad.zero_()
        b.grad.zero_()
    
    # In thông tin mỗi 10 epochs
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1:3d}: Loss = {loss.item():.4f}, w = {w.item():.4f}, b = {b.item():.4f}")

print(f"\n5. Kết quả cuối cùng:")
print(f"w = {w.item():.4f} (mong đợi: 3.0)")
print(f"b = {b.item():.4f} (mong đợi: 5.0)")
print(f"Loss cuối cùng: {loss.item():.4f}")
```

# BTVN 04

Giải thích 2 trường hợp trên

```python
import torch
import numpy as np

print("TRƯỜNG HỢP 1: torch.from_numpy() - Chia sẻ bộ nhớ")

# Tạo numpy array
arr1 = np.array([1, 2, 3, 4, 5])
print(f"Numpy array ban đầu: {arr1}")

# Chuyển đổi sang tensor bằng from_numpy()
x1 = torch.from_numpy(arr1)
print(f"Tensor sau khi chuyển: {x1}")

# Thay đổi giá trị của numpy array
arr1[0] = 99
print(f"\nSau khi thay đổi arr1[0] = 99:")
print(f"Numpy array: {arr1}")
print(f"Tensor:      {x1}")

print("\n=> Tensor cũng thay đổi")
print("   torch.from_numpy() TẠO TENSOR CHIA SẺ BỘ NHỚ với numpy array")
print("   Khi array thay đổi -> tensor cũng thay đổi theo\n")


print("TRƯỜNG HỢP 2: torch.tensor() - Tạo bản sao")

# Tạo numpy array mới
arr2 = np.arange(0, 5)
print(f"Numpy array ban đầu: {arr2}")

# Chuyển đổi sang tensor bằng tensor()
x2 = torch.tensor(arr2)
print(f"Tensor sau khi chuyển: {x2}")

# Thay đổi giá trị của numpy array
arr2[0] = 99
print(f"\nSau khi thay đổi arr2[0] = 99:")
print(f"Numpy array: {arr2}")
print(f"Tensor:      {x2}")

print("\n=> Tensor KHÔNG thay đổi")
print("   torch.tensor() TẠO BẢN SAO ĐỘC LẬP trong bộ nhớ")
print("   Khi array thay đổi -> tensor vẫn giữ nguyên giá trị ban đầu\n")

print("KẾT LUẬN")
print("torch.from_numpy(): Chia sẻ bộ nhớ (memory sharing)")
print("                    + Ưu điểm: Tiết kiệm bộ nhớ, nhanh")
print("                    + Nhược điểm: Thay đổi array ảnh hưởng tensor")
print()
print("torch.tensor():     Tạo bản sao (memory copy)")
print("                    + Ưu điểm: An toàn, độc lập")
print("                    + Nhược điểm: Tốn thêm bộ nhớ")
```

# BTVN 05

Tạo tensor với
 + empty
 + zeros
 + ones
 + random
 + Reshape với view và view as




```python
import torch

print("1. TẠO TENSOR VỚI EMPTY")
empty_tensor = torch.empty(3, 4)
print(f"Shape: {empty_tensor.shape}")
print(empty_tensor)
print()

print("2. TẠO TENSOR VỚI ZEROS")
zeros_tensor = torch.zeros(2, 3, 4)
print(f"Shape: {zeros_tensor.shape}")
print(zeros_tensor)
print()

print("3. TẠO TENSOR VỚI ONES")
ones_tensor = torch.ones(4, 3)
print(f"Shape: {ones_tensor.shape}")
print(ones_tensor)
print()

print("4. TẠO TENSOR VỚI RANDOM")
random_tensor = torch.rand(3, 3)
print(f"Shape: {random_tensor.shape}")
print(random_tensor)
print()

print("5. RESHAPE VỚI VIEW")
original = torch.arange(12)
print(f"Tensor gốc: {original}")
print(f"Shape: {original.shape}")

reshaped = original.view(3, 4)
print(f"\nSau khi view(3, 4):")
print(reshaped)
print(f"Shape: {reshaped.shape}")
print()

print("6. RESHAPE VỚI VIEW_AS")
original2 = torch.arange(24)
print(f"Tensor gốc: shape {original2.shape}")
print(original2)

target_shape = torch.zeros(4, 6)
print(f"\nTarget shape: {target_shape.shape}")

reshaped_as = original2.view_as(target_shape)
print(f"\nSau khi view_as(target_shape):")
print(reshaped_as)
print(f"Shape: {reshaped_as.shape}")
```
